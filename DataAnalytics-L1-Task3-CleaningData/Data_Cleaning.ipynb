```python
import pandas as pd
import numpy as np

print("Pandas version:", pd.__version__)
print("NumPy version:", np.__version__)
print("Libraries imported successfully!")
```

```python
data = {
    "Customer_ID": [101, 102, 103, 104, 105, 105, 106, 107, 108, 109],
    "Name": ["Rahul", "Priya", "Arun", "Sneha", "Kiran", "Kiran", "Vijay", "Anu", "Ravi", "Meena"],
    "Gender": ["Male", "female", "M", "Female", "male", "male", "M", "F", "Male", "female"],
    "Age": [25, 30, 28, np.nan, 35, 35, 29, 150, 32, 27],
    "Purchase_Amount": [1200, 2500, 1800, 2200, np.nan, 3000, 1500, 2000, 99999, 1700],
    "Purchase_Date": [
        "2026-01-10",
        "10/02/2026",
        "2026-03-15",
        "15-04-2026",
        "2026-05-20",
        "2026-05-20",
        "06/06/2026",
        "2026-07-10",
        "2026-08-12",
        "2026-09-05"
    ]
}

df = pd.DataFrame(data)

df
```

```python
print("Dataset Shape:", df.shape)

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())
```

```python
quality_report = pd.DataFrame({
    "Column": df.columns,
    "Data Type": df.dtypes.astype(str).values,
    "Missing Values": df.isnull().sum().values
})

quality_report
```

```python
df.describe(include="all")
```

```python
duplicate_rows = df[df.duplicated(keep=False)]

print("Number of duplicate rows:", df.duplicated().sum())
duplicate_rows
```

```python
print("Gender values before cleaning:")
print(df["Gender"].value_counts())
```

```python
df["Gender"] = df["Gender"].replace({
    "M": "Male",
    "male": "Male",
    "F": "Female",
    "female": "Female"
})

print("Gender values after cleaning:")
print(df["Gender"].value_counts())
```

```python
median_age = df["Age"].median()

df["Age"] = df["Age"].fillna(median_age)

print("Median age used for imputation:", median_age)
print("Missing Age values after cleaning:", df["Age"].isnull().sum())
```

```python
median_purchase = df["Purchase_Amount"].median()

df["Purchase_Amount"] = df["Purchase_Amount"].fillna(median_purchase)

print("Median purchase amount used:", median_purchase)
print(
    "Missing Purchase_Amount values after cleaning:",
    df["Purchase_Amount"].isnull().sum()
)
```

```python
df["Purchase_Date"] = pd.to_datetime(
    df["Purchase_Date"],
    dayfirst=True,
    errors="coerce"
)

print("Purchase_Date data type:", df["Purchase_Date"].dtype)
print("\nPurchase dates after standardization:")
print(df["Purchase_Date"])
```

```python
invalid_age = df[(df["Age"] < 1) | (df["Age"] > 100)]

print("Invalid age records:")
invalid_age
```

```python
median_age = df.loc[df["Age"] <= 100, "Age"].median()

df.loc[(df["Age"] < 1) | (df["Age"] > 100), "Age"] = median_age

print("Invalid age values after cleaning:")
print(df[(df["Age"] < 1) | (df["Age"] > 100)])
```

```python
Q1 = df["Purchase_Amount"].quantile(0.25)
Q3 = df["Purchase_Amount"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["Purchase_Amount"] < lower_bound) |
    (df["Purchase_Amount"] > upper_bound)
]

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("\nOutlier records:")
outliers
```

```python
df["Purchase_Amount"] = df["Purchase_Amount"].clip(
    lower=lower_bound,
    upper=upper_bound
)

print("Purchase amount after outlier handling:")
print(df["Purchase_Amount"])
```

```python
duplicates_before = df.duplicated().sum()

df = df.drop_duplicates().reset_index(drop=True)

print("Duplicate rows removed:", duplicates_before)
print("Dataset shape after removing duplicates:", df.shape)
```

```python
print("=== Cleaned Dataset Check ===")

print("\nShape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nData types:")
print(df.dtypes)

print("\nCleaned Dataset:")
df
```

```python
after_summary = {
    "Rows": len(df),
    "Columns": len(df.columns),
    "Total Missing Values": df.isnull().sum().sum(),
    "Duplicate Rows": df.duplicated().sum(),
    "Correct Data Types": df.dtypes.notnull().sum()
}

after_summary
```

```python
df.to_csv("cleaned_customer_data.csv", index=False)

print("Cleaned dataset saved successfully!")
```

## Conclusion

The dataset was successfully cleaned and transformed into an analysis-ready format.

The following cleaning operations were performed:

- Missing Age and Purchase Amount values were handled using median imputation.
- Duplicate records were identified and removed.
- Inconsistent Gender values were standardized.
- Purchase Date values were converted to datetime format.
- The unrealistic age value was corrected.
- The Purchase Amount outlier was detected using the IQR method and capped.
- Data types and missing values were checked after cleaning.
- The final cleaned dataset was saved as `cleaned_customer_data.csv`.

The cleaned dataset is now ready for further analysis.
