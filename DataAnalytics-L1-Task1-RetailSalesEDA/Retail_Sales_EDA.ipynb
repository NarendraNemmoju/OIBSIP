```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
url = "https://raw.githubusercontent.com/plotly/datasets/master/supermarket_Sales.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
df.head()
```

```python
numeric_df = df.select_dtypes(include=np.number)

print("=== Mean ===")
print(numeric_df.mean())

print("\n=== Median ===")
print(numeric_df.median())

print("\n=== Mode ===")
print(numeric_df.mode().iloc[0])

print("\n=== Standard Deviation ===")
print(numeric_df.std())
```

```python
print("Numerical columns:")
print(numeric_df.columns.tolist())

print("\nSales-related summary:")
print(numeric_df.describe().T)
```

```python
print("Date-related columns:")

for column in df.columns:
    if "date" in column.lower():
        print(column)
```

```python
df["Date"] = pd.to_datetime(df["Date"])

print("Date column converted successfully.")
print("Data type:", df["Date"].dtype)
```

```python
print("All columns:")
print(df.columns.tolist())
```

```python
monthly_sales = df.groupby(
    df["Date"].dt.to_period("M")
)["Total"].sum()

plt.figure(figsize=(12, 5))
monthly_sales.plot(kind="line", marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

```python
quarterly_sales = df.groupby(
    df["Date"].dt.to_period("Q")
)["Total"].sum()

plt.figure(figsize=(12, 5))
quarterly_sales.plot(kind="line", marker="o")

plt.title("Quarterly Sales Trend")
plt.xlabel("Quarter")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

```python
gender_counts = df["Gender"].value_counts()

plt.figure(figsize=(7, 5))
gender_counts.plot(kind="bar")

plt.title("Customer Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Customers")
plt.xticks(rotation=0)

plt.show()

print("Gender distribution:")
print(gender_counts)
```

```python
category_sales = df.groupby("Product line")["Total"].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 5))
category_sales.plot(kind="bar")

plt.title("Sales by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=45, ha="right")

plt.show()

print("Sales by Product Category:")
print(category_sales)
```

```python
top_product_lines = (
    df.groupby("Product line")["Quantity"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10, 5))
top_product_lines.plot(kind="bar")

plt.title("Top-Selling Product Lines")
plt.xlabel("Product Line")
plt.ylabel("Total Quantity Sold")
plt.xticks(rotation=45, ha="right")

plt.show()

print("Top-selling product lines:")
print(top_product_lines)
```

```python
correlation = numeric_df.corr()

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap of Numerical Variables")
plt.show()
```

```python
payment_sales = df.groupby("Payment")["Total"].sum().sort_values(ascending=False)

plt.figure(figsize=(8, 5))
payment_sales.plot(kind="bar")

plt.title("Total Sales by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)

plt.show()

print("Sales by payment method:")
print(payment_sales)
```

## Observations

### Monthly Sales Trend
The monthly sales chart shows how total sales changed over time. The trend helps identify months with relatively higher and lower sales activity.

### Quarterly Sales Trend
The quarterly sales chart provides a broader view of sales performance by grouping individual months into quarters. This makes it easier to observe overall seasonal patterns.

### Gender Distribution
The gender distribution chart shows the number of customers by gender and helps understand the customer composition of the dataset.

### Product Category Sales
The product category analysis shows differences in total sales across product lines. Categories with higher sales contribute more to the overall revenue.

### Top-Selling Product Lines
The top-selling product-line analysis identifies the categories with the highest quantity sold, helping identify products with stronger customer demand.

### Correlation Heatmap
The correlation heatmap shows relationships between numerical variables. Stronger positive or negative correlation values indicate stronger relationships between variables.

### Payment Method Analysis
The payment-method chart shows how total sales are distributed across different payment methods and provides an additional view of customer purchasing behavior.

## Conclusion and Business Recommendations

The exploratory data analysis provided an overview of sales performance, customer demographics, product categories, and payment behavior.

### Business Recommendations

1. **Focus on high-performing product categories:** Product lines with higher sales and quantity should receive sufficient inventory and marketing attention to maintain their performance.

2. **Plan inventory based on sales trends:** Monthly and quarterly sales trends can be used to prepare inventory and promotions for periods with higher customer demand.

3. **Use customer demographic insights:** The gender distribution can help businesses understand their customer base and design more relevant marketing campaigns.

4. **Monitor payment preferences:** Sales by payment method can help the business understand customer purchasing preferences and ensure convenient payment options are available.

5. **Use data-driven decision making:** Regular analysis of sales, product demand, and customer behavior can help improve inventory planning, marketing, and overall business performance.
